```python
import pandas as pd
import numpy as np
import random

print("Python environment is working!")
```

```python
random.seed(42)
np.random.seed(42)

print("Random seed configured.")
```

```python
num_customers = 50_000

print(f"Number of customers to generate: {num_customers:,}")
```

```python
first_names = [
    "James", "Robert", "John", "Michael", "David",
    "William", "Richard", "Joseph", "Thomas", "Charles",
    "Mary", "Patricia", "Jennifer", "Linda", "Elizabeth",
    "Barbara", "Susan", "Jessica", "Sarah", "Karen"
]

last_names = [
    "Smith", "Johnson", "Williams", "Brown", "Jones",
    "Garcia", "Miller", "Davis", "Rodriguez", "Martinez",
    "Wilson", "Anderson", "Taylor", "Thomas", "Moore",
    "Jackson", "Martin", "Lee", "Perez", "Thompson"
]

print(f"First names available: {len(first_names)}")
print(f"Last names available: {len(last_names)}")
```

```python
customer_first_names = np.random.choice(
    first_names,
    size=num_customers
)

customer_last_names = np.random.choice(
    last_names,
    size=num_customers
)

print("Customer names generated successfully.")
```

```python
genders = np.random.choice(
    ["Male", "Female", "Other"],
    size=num_customers,
    p=[0.48, 0.48, 0.04]
)

occupations = np.random.choice(
    [
        "Engineer",
        "Teacher",
        "Healthcare",
        "Finance",
        "IT",
        "Sales",
        "Manager",
        "Consultant",
        "Government",
        "Self-Employed",
        "Retired"
    ],
    size=num_customers
)

print("Gender and occupation generated successfully.")
```

```python
start_birth = pd.Timestamp("1940-01-01")
end_birth = pd.Timestamp("2008-12-31")

birth_dates = pd.to_datetime(
    np.random.randint(
        start_birth.value // 10**9,
        end_birth.value // 10**9,
        size=num_customers
    ),
    unit="s"
)

print("Date of birth generated successfully.")
```

```python
annual_income = np.random.lognormal(
    mean=10.8,
    sigma=0.55,
    size=num_customers
)

annual_income = np.round(
    np.clip(annual_income, 25_000, 500_000),
    2
)

print("Annual income generated successfully.")
```

```python
cities_states = [
    ("New York", "NY"),
    ("Los Angeles", "CA"),
    ("Chicago", "IL"),
    ("Houston", "TX"),
    ("Phoenix", "AZ"),
    ("Philadelphia", "PA"),
    ("San Antonio", "TX"),
    ("San Diego", "CA"),
    ("Dallas", "TX"),
    ("Austin", "TX"),
    ("Atlanta", "GA"),
    ("Miami", "FL"),
    ("Charlotte", "NC"),
    ("Seattle", "WA"),
    ("Boston", "MA")
]

selected_locations = random.choices(
    cities_states,
    k=num_customers
)

cities = [location[0] for location in selected_locations]
states = [location[1] for location in selected_locations]

print("Customer locations generated successfully.")
```

```python
customer_segments = np.random.choice(
    [
        "Mass Market",
        "Mass Affluent",
        "Affluent",
        "Premium",
        "Business"
    ],
    size=num_customers,
    p=[0.45, 0.25, 0.15, 0.10, 0.05]
)

print("Customer segments generated successfully.")
```

```python
join_start = pd.Timestamp("2020-01-01")
join_end = pd.Timestamp("2026-07-01")

join_dates = pd.to_datetime(
    np.random.randint(
        join_start.value // 10**9,
        join_end.value // 10**9,
        size=num_customers
    ),
    unit="s"
)

print("Customer join dates generated successfully.")
```

```python
risk_categories = np.random.choice(
    [
        "Low",
        "Medium",
        "High"
    ],
    size=num_customers,
    p=[0.60, 0.30, 0.10]
)

print("Risk categories generated successfully.")
```

```python
customers = pd.DataFrame({
    "CustomerID": range(1, num_customers + 1),
    "FirstName": customer_first_names,
    "LastName": customer_last_names,
    "DateOfBirth": birth_dates,
    "Gender": genders,
    "Occupation": occupations,
    "AnnualIncome": annual_income,
    "City": cities,
    "State": states,
    "CustomerSegment": customer_segments,
    "JoinDate": join_dates,
    "RiskCategory": risk_categories,
    "Status": "Active"
})

print("Customer DataFrame created successfully.")
```
