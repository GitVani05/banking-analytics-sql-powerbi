```python
import pandas as pd
import numpy as np
import random

print("Python environment is working!")
```

```python
random.seed(42)
np.random.seed(42)

print("Random seed configured.")
```

```python
cities_states = [
    ("New York", "NY"),
    ("Los Angeles", "CA"),
    ("Chicago", "IL"),
    ("Houston", "TX"),
    ("Phoenix", "AZ"),
    ("Philadelphia", "PA"),
    ("San Antonio", "TX"),
    ("San Diego", "CA"),
    ("Dallas", "TX"),
    ("Austin", "TX"),
    ("Atlanta", "GA"),
    ("Miami", "FL"),
    ("Charlotte", "NC"),
    ("Seattle", "WA"),
    ("Boston", "MA")
]

regions = {
    "NY": "Northeast",
    "CA": "West",
    "IL": "Midwest",
    "TX": "South",
    "AZ": "West",
    "PA": "Northeast",
    "FL": "South",
    "NC": "South",
    "WA": "West",
    "MA": "Northeast",
    "GA": "South"
}

num_branches = 100

selected_locations = random.choices(
    cities_states,
    k=num_branches
)

branch_names = [
    f"{city} Branch {i:03d}"
    for i, (city, state) in enumerate(selected_locations, start=1)
]

branch_cities = [location[0] for location in selected_locations]
branch_states = [location[1] for location in selected_locations]
branch_regions = [regions[state] for state in branch_states]

print("Branches generated successfully.")
```

```python
branch_opening_dates = pd.to_datetime(
    np.random.choice(
        pd.date_range(
            start="2000-01-01",
            end="2025-12-31",
            freq="D"
        ),
        size=num_branches,
        replace=False
    )
)

manager_first_names = [
    "James", "Robert", "Michael", "David", "William",
    "Linda", "Patricia", "Jennifer", "Sarah", "Karen"
]

manager_last_names = [
    "Smith", "Johnson", "Brown", "Davis", "Wilson",
    "Taylor", "Anderson", "Thomas", "Moore", "Martin"
]

manager_names = [
    f"{random.choice(manager_first_names)} "
    f"{random.choice(manager_last_names)}"
    for _ in range(num_branches)
]
```

```python
branches = pd.DataFrame({
    "BranchID": range(1, num_branches + 1),
    "BranchName": branch_names,
    "City": branch_cities,
    "State": branch_states,
    "Region": branch_regions,
    "OpeningDate": branch_opening_dates,
    "ManagerName": manager_names,
    "Status": "Active"
})

branches.head()
```

```python
branch_output_path = "../data/raw/DimBranch.csv"

branches.to_csv(
    branch_output_path,
    index=False
)

print(f"Saved {len(branches)} branches to {branch_output_path}")
```

```python
num_accounts = 75_000

account_customer_ids = np.random.randint(
    1,
    50_001,
    size=num_accounts
)
```

```python
account_types = np.random.choice(
    [
        "Checking",
        "Savings",
        "Money Market",
        "Business Checking",
        "Business Savings"
    ],
    size=num_accounts,
    p=[0.40, 0.35, 0.10, 0.10, 0.05]
)
```

```python
account_open_dates = pd.to_datetime(
    np.random.choice(
        pd.date_range(
            start="2020-01-01",
            end="2026-07-01",
            freq="D"
        ),
        size=num_accounts,
        replace=True
    )
)
```

```python
account_balances = np.round(
    np.random.lognormal(
        mean=8.5,
        sigma=1.1,
        size=num_accounts
    ),
    2
)

account_balances = np.clip(
    account_balances,
    0,
    1_000_000
)
```

```python
account_statuses = np.random.choice(
    [
        "Active",
        "Inactive",
        "Closed"
    ],
    size=num_accounts,
    p=[0.90, 0.07, 0.03]
)
```

```python
account_numbers = [
    f"ACCT-{i:08d}"
    for i in range(1, num_accounts + 1)
]
```

```python
account_branch_ids = np.random.randint(
    1,
    num_branches + 1,
    size=num_accounts
)
```

```python
accounts = pd.DataFrame({
    "AccountID": range(1, num_accounts + 1),
    "CustomerID": account_customer_ids,
    "AccountNumber": account_numbers,
    "AccountType": account_types,
    "BranchID": account_branch_ids,
    "OpenDate": account_open_dates,
    "CurrentBalance": account_balances,
    "Status": account_statuses
})

accounts.head()
```
